```python

import pandas as pd
import numpy as np
from pathlib import Path

# Step 1: Set random seed
rng = np.random.default_rng(42)

# Step 2: Define supermarket products and prices
products = {
    "Rice": ("Groceries", 800),
    "Milk": ("Groceries", 60),
    "Bread": ("Groceries", 45),
    "Cooking Oil": ("Groceries", 180),
    "T-Shirt": ("Clothing", 499),
    "Jeans": ("Clothing", 1299),
    "Earphones": ("Electronics", 850),
    "Power Bank": ("Electronics", 1400),
    "Shampoo": ("Beauty", 250),
    "Detergent": ("Home", 220)
}

# Step 3: Generate 2,000 transactions
records = []
product_names = list(products.keys())

for i in range(2000):
    product = rng.choice(product_names)
    category, price = products[product]

    quantity = int(rng.integers(1, 6))
    discount = int(rng.choice([0, 5, 10, 15]))

    sales = round(quantity * price * (1 - discount / 100), 2)

    records.append({
        "TransactionID": i + 1,
        "Date": str(
            pd.Timestamp("2026-01-01")
            + pd.Timedelta(days=int(rng.integers(0, 90)))
        )[:10],
        "CustomerID": int(rng.integers(1001, 1401)),
        "Product": product,
        "Category": category,
        "Quantity": quantity,
        "UnitPrice": price,
        "Discount": discount,
        "Sales": sales,
        "PaymentMethod": rng.choice(
            ["UPI", "Cash", "Card"]
        )
    })

# Step 4: Convert to Pandas DataFrame
df = pd.DataFrame(records)

# Step 5: Locate the project data folder
project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

data_folder = project_root / "data"
data_folder.mkdir(parents=True, exist_ok=True)

# Step 6: Save as CSV
csv_path = data_folder / "supermarket_sales.csv"
df.to_csv(csv_path, index=False)

print("Dataset created successfully!")
print("Saved to:", csv_path)
print("Dataset shape:", df.shape)

# Display first five rows
display(df.head())
```

```python

import pandas as pd
import numpy as np
from pathlib import Path

# Find the project folder
project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

# Load the original CSV
file_path = project_root / "data" / "supermarket_sales.csv"

df_original = pd.read_csv(file_path)

# Display dataset information
print("Original shape:", df_original.shape)
display(df_original.head())

print("\nMissing values:")
print(df_original.isnull().sum())

print("\nDuplicate records:")
print(df_original.duplicated().sum())
```

```python

# Create an independent copy
df_dirty = df_original.copy()

# Allow mixed data types for our practice
df_dirty["Quantity"] = df_dirty["Quantity"].astype(object)

# Introduce common data quality issues
df_dirty.loc[0, "Product"] = " rice "
df_dirty.loc[1, "Quantity"] = "not available"
df_dirty.loc[2, "Discount"] = 150
df_dirty.loc[3, "Sales"] = 999999
df_dirty.loc[4, "Date"] = "not-a-date"
df_dirty.loc[5, "PaymentMethod"] = None
df_dirty.loc[7, "UnitPrice"] = -50
df_dirty.loc[8, "Category"] = " groceries "
df_dirty.loc[10, "Quantity"] = -2

# Duplicate one transaction
duplicate_row = df_dirty.iloc[[6]].copy()

df_dirty = pd.concat(
    [df_dirty, duplicate_row],
    ignore_index=True
)

print("Messy dataset shape:", df_dirty.shape)
display(df_dirty.head(12))
```

```python

print("Missing values in messy dataset:")
print(df_dirty.isnull().sum())

print("\nMissing values by percentage:")

missing_percentage = (
    df_dirty.isnull().sum()
    / len(df_dirty)
) * 100

print(missing_percentage.round(2))
```

```python

df_working = df_dirty.copy()

# Clean product names
df_working["Product"] = (
    df_working["Product"]
    .astype("string")
    .str.strip()
    .str.title()
)

# Clean category names
df_working["Category"] = (
    df_working["Category"]
    .astype("string")
    .str.strip()
    .str.title()
)

# Clean payment method names
df_working["PaymentMethod"] = (
    df_working["PaymentMethod"]
    .astype("string")
    .str.strip()
    .str.upper()
    .fillna("UNKNOWN")
)

print("Text standardization completed!")
display(df_working.head())
```

```python

numeric_columns = [
    "TransactionID",
    "CustomerID",
    "Quantity",
    "UnitPrice",
    "Discount",
    "Sales"
]

for column in numeric_columns:
    df_working[column] = pd.to_numeric(
        df_working[column],
        errors="coerce"
    )

print(df_working[numeric_columns].dtypes)

print("\nMissing quantities after conversion:")
print(df_working["Quantity"].isna().sum())
```

```python

df_working["Date"] = pd.to_datetime(
    df_working["Date"],
    format="%Y-%m-%d",
    errors="coerce"
)

print("Date column type:")
print(df_working["Date"].dtype)

print("\nInvalid or missing dates:")
print(df_working["Date"].isna().sum())
```

```python

# Essential columns must not be missing
required_columns = [
    "TransactionID",
    "Date",
    "CustomerID",
    "Product",
    "Category",
    "Quantity",
    "UnitPrice",
    "Discount"
]

# Build a mask of invalid records
invalid_mask = (
    df_working[required_columns].isna().any(axis=1)
    | (df_working["TransactionID"] <= 0)
    | (df_working["CustomerID"] <= 0)
    | (df_working["Quantity"] <= 0)
    | (df_working["Quantity"] % 1 != 0)
    | (df_working["UnitPrice"] <= 0)
    | ~df_working["Discount"].between(0, 100)
)

# Keep rejected records for auditing
df_rejected = df_working.loc[invalid_mask].copy()

# Keep valid records
df_valid = df_working.loc[~invalid_mask].copy()

print("Rejected rows:", len(df_rejected))
print("Remaining rows:", len(df_valid))
```

```python

# Identify repeated transaction IDs
duplicate_mask = df_valid.duplicated(
    subset=["TransactionID"],
    keep="first"
)

print("Duplicate IDs found:", duplicate_mask.sum())

# Remove duplicate IDs
df_clean = df_valid.loc[~duplicate_mask].copy()

# Reset row index
df_clean.reset_index(drop=True, inplace=True)

print("Rows after removing duplicates:", len(df_clean))
```

```python

# Calculate expected sales
expected_sales = (
    df_clean["Quantity"]
    * df_clean["UnitPrice"]
    * (1 - df_clean["Discount"] / 100)
).round(2)

# Find discrepancies
sales_mismatch = (
    df_clean["Sales"].isna()
    | ~np.isclose(
        df_clean["Sales"],
        expected_sales,
        atol=0.01,
        rtol=0
    )
)

print("Incorrect sales values:", sales_mismatch.sum())

# Correct sales using the source fields
df_clean["Sales"] = expected_sales

print("Sales values recalculated successfully!")
```

```python

# Convert integer-like columns back to integers
for column in ["TransactionID", "CustomerID", "Quantity"]:
    df_clean[column] = df_clean[column].astype("int64")

# Check cleaned data
print("Final shape:", df_clean.shape)

print("Missing values:")
print(df_clean.isna().sum())

print("Duplicate transaction IDs:",
      df_clean["TransactionID"].duplicated().sum())

print("Total Sales:", round(df_clean["Sales"].sum(), 2))

# Automated data quality checks
assert df_clean["TransactionID"].is_unique
assert df_clean["Quantity"].gt(0).all()
assert df_clean["UnitPrice"].gt(0).all()
assert df_clean["Discount"].between(0, 100).all()
assert df_clean["Date"].notna().all()

print("All quality checks passed!")
```

```python

# Save the cleaned dataset
output_path = (
    project_root
    / "data"
    / "supermarket_sales_practice_cleaned.csv"
)

df_clean.to_csv(output_path, index=False)

print("Cleaned dataset saved!")
print("Location:", output_path)
```

```python

# ============================================================
# STEP 8: COMPLETE STATISTICAL ANALYSIS
# PROJECT: SUPERMARKET SALES ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from IPython.display import display

# ------------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------------

project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

file_path = project_root / "data" / "supermarket_sales.csv"

df = pd.read_csv(file_path, parse_dates=["Date"])

print("=" * 60)
print("STEP 8: SUPERMARKET STATISTICAL ANALYSIS")
print("=" * 60)

print("\nDataset Shape:", df.shape)
print("Date Range:", df["Date"].min().date(),
      "to", df["Date"].max().date())

# ------------------------------------------------------------
# 2. DESCRIPTIVE STATISTICS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("1. DESCRIPTIVE STATISTICS")
print("=" * 60)

sales = df["Sales"]

mean_sales = sales.mean()
median_sales = sales.median()
mode_sales = sales.mode().iloc[0]
std_sales = sales.std()
variance_sales = sales.var()
min_sales = sales.min()
max_sales = sales.max()
total_sales = sales.sum()

print(f"Total Revenue       : ₹{total_sales:,.2f}")
print(f"Mean Sales          : ₹{mean_sales:,.2f}")
print(f"Median Sales        : ₹{median_sales:,.2f}")
print(f"Mode Sales          : ₹{mode_sales:,.2f}")
print(f"Standard Deviation  : ₹{std_sales:,.2f}")
print(f"Variance            : {variance_sales:,.2f}")
print(f"Minimum Sales       : ₹{min_sales:,.2f}")
print(f"Maximum Sales       : ₹{max_sales:,.2f}")
print(f"Range               : ₹{max_sales - min_sales:,.2f}")

print("\nStatistical Summary:")
display(sales.describe().to_frame("Sales"))

# ------------------------------------------------------------
# 3. QUARTILES AND INTERQUARTILE RANGE (IQR)
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("2. QUARTILES AND IQR")
print("=" * 60)

q1 = sales.quantile(0.25)
q2 = sales.quantile(0.50)
q3 = sales.quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print(f"Q1 (25th Percentile): ₹{q1:,.2f}")
print(f"Q2 (Median)         : ₹{q2:,.2f}")
print(f"Q3 (75th Percentile): ₹{q3:,.2f}")
print(f"IQR                 : ₹{iqr:,.2f}")
print(f"Lower Bound         : ₹{lower_bound:,.2f}")
print(f"Upper Bound         : ₹{upper_bound:,.2f}")

# ------------------------------------------------------------
# 4. OUTLIER DETECTION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("3. OUTLIER DETECTION")
print("=" * 60)

outliers = df[
    (df["Sales"] < lower_bound) |
    (df["Sales"] > upper_bound)
]

outlier_percentage = len(outliers) / len(df) * 100

print("Total Outliers:", len(outliers))
print(f"Outlier Percentage: {outlier_percentage:.2f}%")

print("\nSample Outlier Records:")
display(outliers.head(10))

# NOTE: Outliers are not automatically errors.
# High-value purchases may be genuine.

# ------------------------------------------------------------
# 5. CATEGORY-WISE STATISTICAL ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("4. CATEGORY-WISE STATISTICS")
print("=" * 60)

category_stats = (
    df.groupby("Category")
    .agg(
        RecordCount=("Sales", "count"),
        TotalRevenue=("Sales", "sum"),
        MeanSales=("Sales", "mean"),
        MedianSales=("Sales", "median"),
        StdDeviation=("Sales", "std"),
        MinimumSales=("Sales", "min"),
        MaximumSales=("Sales", "max"),
        TotalQuantity=("Quantity", "sum")
    )
    .sort_values("TotalRevenue", ascending=False)
)

display(category_stats.round(2))

print("\nHighest Revenue Category:",
      category_stats.index[0])

# ------------------------------------------------------------
# 6. CORRELATION ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("5. CORRELATION ANALYSIS")
print("=" * 60)

numeric_columns = [
    "Quantity",
    "UnitPrice",
    "Discount",
    "Sales"
]

correlation_matrix = df[numeric_columns].corr()

print("\nPearson Correlation Matrix:")
display(correlation_matrix.round(3))

print("\nCorrelation with Sales:")
print(
    correlation_matrix["Sales"]
    .sort_values(ascending=False)
    .round(3)
)

# Correlation does not establish causation.
# These are simulated product-line transactions.

# ------------------------------------------------------------
# 7. DAILY SALES TREND
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("6. DAILY SALES TREND")
print("=" * 60)

daily_sales = (
    df.groupby("Date")["Sales"]
    .sum()
    .sort_index()
)

# Include all dates in the observed period
all_dates = pd.date_range(
    df["Date"].min(),
    df["Date"].max(),
    freq="D"
)

daily_sales = daily_sales.reindex(
    all_dates, fill_value=0
)

daily_sales.index.name = "Date"

# Seven-day moving average
moving_average = daily_sales.rolling(
    window=7,
    min_periods=1
).mean()

print("Average Daily Revenue:",
      round(daily_sales.mean(), 2))

print("Highest Revenue Day:",
      daily_sales.idxmax().date())

print("Highest Daily Revenue:",
      round(daily_sales.max(), 2))

print("Lowest Revenue Day:",
      daily_sales.idxmin().date())

# Estimated linear trend in rupees per day
x = np.arange(len(daily_sales))

trend_slope = np.polyfit(
    x,
    daily_sales.values,
    1
)[0]

print(f"Linear Trend Slope: ₹{trend_slope:,.2f} per day")

# ------------------------------------------------------------
# 8. MONTHLY SALES TREND
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("7. MONTHLY SALES TREND")
print("=" * 60)

df["Month"] = df["Date"].dt.to_period("M").astype(str)

monthly_sales = (
    df.groupby("Month")
    .agg(
        TotalRevenue=("Sales", "sum"),
        Records=("TransactionID", "nunique")
    )
    .sort_index()
)

monthly_sales["GrowthPercent"] = (
    monthly_sales["TotalRevenue"].pct_change() * 100
)

display(monthly_sales.round(2))

print("\nBest Revenue Month:",
      monthly_sales["TotalRevenue"].idxmax())

# ------------------------------------------------------------
# 9. WEEKDAY SALES ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("8. WEEKDAY ANALYSIS")
print("=" * 60)

daily_summary = daily_sales.rename(
    "DailyRevenue"
).reset_index()

daily_summary["Weekday"] = (
    daily_summary["Date"].dt.day_name()
)

weekday_sales = (
    daily_summary.groupby("Weekday")
    .agg(
        AverageDailyRevenue=("DailyRevenue", "mean"),
        TotalRevenue=("DailyRevenue", "sum"),
        DaysObserved=("DailyRevenue", "size")
    )
)

weekday_order = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

weekday_sales = weekday_sales.reindex(weekday_order)

display(weekday_sales.round(2))

print("Highest Average Daily Revenue Weekday:",
      weekday_sales["AverageDailyRevenue"].idxmax())

# ------------------------------------------------------------
# 10. STATISTICAL VISUALIZATIONS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("9. STATISTICAL VISUALIZATIONS")
print("=" * 60)

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(
    2, 2, figsize=(15, 10)
)

# Chart 1: Sales distribution
sns.histplot(
    df["Sales"],
    bins=30,
    kde=True,
    ax=axes[0, 0]
)

axes[0, 0].axvline(
    mean_sales,
    linestyle="--",
    label=f"Mean ₹{mean_sales:,.0f}"
)

axes[0, 0].axvline(
    median_sales,
    linestyle=":",
    label=f"Median ₹{median_sales:,.0f}"
)

axes[0, 0].set_title("Distribution of Sales")
axes[0, 0].set_xlabel("Sales (₹)")
axes[0, 0].legend()

# Chart 2: Box plot for outlier analysis
sns.boxplot(
    x=df["Sales"],
    ax=axes[0, 1]
)

axes[0, 1].set_title("Sales Outlier Analysis")
axes[0, 1].set_xlabel("Sales (₹)")

# Chart 3: Daily sales trend
axes[1, 0].plot(
    daily_sales.index,
    daily_sales.values,
    label="Daily Revenue",
    alpha=0.5
)

axes[1, 0].plot(
    moving_average.index,
    moving_average.values,
    label="7-Day Moving Average",
    linewidth=2
)

axes[1, 0].set_title("Daily Revenue Trend")
axes[1, 0].set_xlabel("Date")
axes[1, 0].set_ylabel("Revenue (₹)")
axes[1, 0].legend()
axes[1, 0].tick_params(axis="x", rotation=30)

# Chart 4: Correlation heatmap
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    ax=axes[1, 1]
)

axes[1, 1].set_title("Correlation Heatmap")

plt.tight_layout()

# Save visualization
visual_dir = project_root / "visualizations"
visual_dir.mkdir(parents=True, exist_ok=True)

chart_path = visual_dir / "step8_statistics.png"

plt.savefig(
    chart_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------------------
# 11. SAVE STATISTICAL REPORT TABLES
# ------------------------------------------------------------

reports_dir = project_root / "reports"
reports_dir.mkdir(parents=True, exist_ok=True)

category_stats.to_csv(
    reports_dir / "category_statistics.csv"
)

correlation_matrix.to_csv(
    reports_dir / "correlation_matrix.csv"
)

monthly_sales.to_csv(
    reports_dir / "monthly_statistics.csv"
)

weekday_sales.to_csv(
    reports_dir / "weekday_statistics.csv"
)

# ------------------------------------------------------------
# 12. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FINAL STATISTICAL ANALYSIS SUMMARY")
print("=" * 60)

print(f"Total Records: {len(df):,}")
print(f"Total Revenue: ₹{total_sales:,.2f}")
print(f"Mean Sales: ₹{mean_sales:,.2f}")
print(f"Median Sales: ₹{median_sales:,.2f}")
print(f"Standard Deviation: ₹{std_sales:,.2f}")
print(f"Sales IQR: ₹{iqr:,.2f}")
print(f"Outliers Detected: {len(outliers)}")
print(f"Top Revenue Category: {category_stats.index[0]}")
print(f"Best Revenue Month: {monthly_sales['TotalRevenue'].idxmax()}")

print("\nImportant Interpretation:")
print("- Statistics describe simulated product-line sales.")
print("- Outliers should be investigated, not automatically removed.")
print("- Correlation does not prove cause and effect.")
print("- Trends in simulated data are not business forecasts.")

print("\nReports saved in:", reports_dir)
print("Visualization saved at:", chart_path)

print("\nSTEP 8 COMPLETED SUCCESSFULLY!")
```

```python

# ============================================================
# STEP 10: BUSINESS INSIGHTS AND PROFESSIONAL PDF REPORT
# PROJECT: SUPERMARKET SALES ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path
from datetime import datetime
from xml.sax.saxutils import escape

from reportlab.platypus import (
    SimpleDocTemplate, Paragraph, Spacer, Table,
    TableStyle, Image, PageBreak, KeepTogether
)
from reportlab.lib import colors
from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.enums import TA_CENTER

# ------------------------------------------------------------
# 1. CONFIGURE PROJECT PATHS
# ------------------------------------------------------------

project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

data_path = project_root / "data" / "supermarket_sales.csv"
chart_dir = project_root / "visualizations" / "step9"
reports_dir = project_root / "reports"

reports_dir.mkdir(parents=True, exist_ok=True)

pdf_path = reports_dir / "supermarket_sales_report.pdf"
insights_path = reports_dir / "business_insights.md"

# ------------------------------------------------------------
# 2. LOAD AND VALIDATE DATA
# ------------------------------------------------------------

df = pd.read_csv(data_path, parse_dates=["Date"])

required_columns = [
    "TransactionID", "Date", "CustomerID",
    "Product", "Category", "Quantity",
    "UnitPrice", "Discount", "Sales"
]

missing_columns = set(required_columns) - set(df.columns)

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

if df.empty:
    raise ValueError("The dataset is empty.")

print("=" * 60)
print("STEP 10: BUSINESS REPORT GENERATION")
print("=" * 60)

# ------------------------------------------------------------
# 3. CALCULATE KEY PERFORMANCE METRICS
# ------------------------------------------------------------

total_records = len(df)
unique_customers = df["CustomerID"].nunique()
total_revenue = df["Sales"].sum()
total_units = df["Quantity"].sum()

average_sales = df["Sales"].mean()
median_sales = df["Sales"].median()

gross_sales = (
    df["Quantity"] * df["UnitPrice"]
).sum()

total_discount = gross_sales - total_revenue

discount_pct = (
    total_discount / gross_sales * 100
    if gross_sales > 0 else 0
)

start_date = df["Date"].min().strftime("%d %b %Y")
end_date = df["Date"].max().strftime("%d %b %Y")

# ------------------------------------------------------------
# 4. CATEGORY ANALYSIS
# ------------------------------------------------------------

category_summary = (
    df.groupby("Category", as_index=False)
    .agg(
        Revenue=("Sales", "sum"),
        Units=("Quantity", "sum"),
        Records=("TransactionID", "nunique")
    )
    .sort_values("Revenue", ascending=False)
)

category_summary["SharePercent"] = (
    category_summary["Revenue"] / total_revenue * 100
)

best_category = category_summary.iloc[0]

# ------------------------------------------------------------
# 5. PRODUCT ANALYSIS
# ------------------------------------------------------------

product_summary = (
    df.groupby("Product", as_index=False)
    .agg(
        Revenue=("Sales", "sum"),
        Units=("Quantity", "sum")
    )
    .sort_values("Revenue", ascending=False)
)

best_product = product_summary.iloc[0]

best_unit_product = product_summary.sort_values(
    "Units", ascending=False
).iloc[0]

# ------------------------------------------------------------
# 6. MONTHLY SALES ANALYSIS
# ------------------------------------------------------------

df["Month"] = df["Date"].dt.to_period("M").astype(str)

monthly_summary = (
    df.groupby("Month", as_index=False)
    .agg(
        Revenue=("Sales", "sum"),
        Records=("TransactionID", "nunique")
    )
    .sort_values("Month")
)

monthly_summary["GrowthPercent"] = (
    monthly_summary["Revenue"].pct_change() * 100
)

best_month = monthly_summary.loc[
    monthly_summary["Revenue"].idxmax()
]

# ------------------------------------------------------------
# 7. GENERATE BUSINESS INSIGHTS
# ------------------------------------------------------------

insights = [
    (
        f"The dataset contains {total_records:,} simulated "
        f"product-line sales records from {start_date} "
        f"to {end_date}."
    ),
    (
        f"Recorded sales after discounts total "
        f"INR {total_revenue:,.2f}, with "
        f"{total_units:,} units sold."
    ),
    (
        f"{best_category['Category']} is the highest-revenue "
        f"category, contributing "
        f"{best_category['SharePercent']:.1f}% "
        f"of recorded sales."
    ),
    (
        f"{best_product['Product']} generates the highest "
        f"product revenue at INR "
        f"{best_product['Revenue']:,.2f}."
    ),
    (
        f"{best_unit_product['Product']} has the highest "
        f"number of units sold, with "
        f"{best_unit_product['Units']:,.0f} units."
    ),
    (
        f"{best_month['Month']} is the highest-revenue "
        f"month, with INR {best_month['Revenue']:,.2f}."
    ),
    (
        f"Average product-line sales equal "
        f"INR {average_sales:,.2f}, compared with "
        f"median sales of INR {median_sales:,.2f}."
    )
]

# ------------------------------------------------------------
# 8. BUSINESS RECOMMENDATIONS
# ------------------------------------------------------------

recommendations = [
    (
        f"Review inventory planning for "
        f"{best_product['Product']} and the "
        f"{best_category['Category']} category. "
        f"Verify demand and stockout information before "
        f"changing stock levels."
    ),
    (
        "Compare revenue with cost and profit-margin data "
        "before deciding which categories deserve more "
        "investment."
    ),
    (
        "Investigate monthly revenue differences using "
        "promotions, holidays, operating days, and "
        "availability data before attributing changes "
        "to a particular cause."
    ),
    (
        "Study product-line sales distributions to "
        "identify unusually large purchases and verify "
        "whether they are genuine."
    ),
    (
        "Track category revenue, quantity sold, and "
        "discount amounts in a recurring dashboard "
        "when real operational data is available."
    )
]

# ------------------------------------------------------------
# 9. DISPLAY FINDINGS IN THE NOTEBOOK
# ------------------------------------------------------------

print("\nEXECUTIVE SUMMARY")
print("-" * 60)

print(f"Records             : {total_records:,}")
print(f"Unique Customers    : {unique_customers:,}")
print(f"Recorded Sales      : INR {total_revenue:,.2f}")
print(f"Gross Sales         : INR {gross_sales:,.2f}")
print(f"Discount Amount     : INR {total_discount:,.2f}")
print(f"Discount Share      : {discount_pct:.2f}%")
print(f"Units Sold          : {total_units:,}")
print(f"Average Line Sales  : INR {average_sales:,.2f}")
print(f"Median Line Sales   : INR {median_sales:,.2f}")

print("\nBUSINESS INSIGHTS")
for i, insight in enumerate(insights, 1):
    print(f"{i}. {insight}")

print("\nRECOMMENDATIONS")
for i, recommendation in enumerate(recommendations, 1):
    print(f"{i}. {recommendation}")

# ------------------------------------------------------------
# 10. SAVE BUSINESS INSIGHTS AS MARKDOWN
# ------------------------------------------------------------

markdown_lines = [
    "# Supermarket Sales - Business Insights",
    "",
    "**Dataset:** Simulated supermarket sales",
    f"**Period:** {start_date} to {end_date}",
    "",
    "## Executive Summary",
    "",
    f"- Records: {total_records:,}",
    f"- Recorded sales: INR {total_revenue:,.2f}",
    f"- Units sold: {total_units:,}",
    f"- Unique customer IDs: {unique_customers:,}",
    "",
    "## Key Findings",
    ""
]

markdown_lines += [
    f"{i}. {item}"
    for i, item in enumerate(insights, 1)
]

markdown_lines += [
    "",
    "## Recommendations",
    ""
]

markdown_lines += [
    f"{i}. {item}"
    for i, item in enumerate(recommendations, 1)
]

markdown_lines += [
    "",
    "## Limitations",
    "",
    "- The dataset is simulated, not real company data.",
    "- Each record represents one product line.",
    "- Revenue does not establish profitability.",
    "- Observed patterns do not establish causation.",
    "- No hourly or promotional-effect analysis is possible.",
]

insights_path.write_text(
    "\n".join(markdown_lines),
    encoding="utf-8"
)

# ------------------------------------------------------------
# 11. CONFIGURE PROFESSIONAL PDF STYLES
# ------------------------------------------------------------

styles = getSampleStyleSheet()

styles.add(ParagraphStyle(
    name="ReportTitle",
    parent=styles["Title"],
    fontSize=22,
    leading=27,
    textColor=colors.HexColor("#17365D"),
    spaceAfter=16
))

styles.add(ParagraphStyle(
    name="ReportSection",
    parent=styles["Heading2"],
    fontSize=13,
    leading=17,
    textColor=colors.HexColor("#17365D"),
    spaceBefore=14,
    spaceAfter=8
))

styles.add(ParagraphStyle(
    name="ReportBody",
    parent=styles["BodyText"],
    fontSize=9,
    leading=14,
    spaceAfter=8
))

styles.add(ParagraphStyle(
    name="ReportNote",
    parent=styles["Normal"],
    fontSize=8,
    leading=11,
    textColor=colors.HexColor("#64748B")
))

def para(value, style="ReportBody"):
    return Paragraph(escape(str(value)), styles[style])

def section(title):
    return para(title, "ReportSection")

def make_table(rows, widths):
    table = Table(
        rows,
        colWidths=widths,
        repeatRows=1,
        hAlign="LEFT"
    )

    table.setStyle(TableStyle([
        (
            "BACKGROUND", (0, 0), (-1, 0),
            colors.HexColor("#17365D")
        ),
        (
            "TEXTCOLOR", (0, 0), (-1, 0),
            colors.white
        ),
        (
            "FONTNAME", (0, 0), (-1, 0),
            "Helvetica-Bold"
        ),
        ("FONTSIZE", (0, 0), (-1, -1), 8),
        ("BOTTOMPADDING", (0, 0), (-1, -1), 8),
        ("TOPPADDING", (0, 0), (-1, -1), 8),
        (
            "ROWBACKGROUNDS", (0, 1), (-1, -1),
            [colors.white, colors.HexColor("#F1F5F9")]
        ),
        ("VALIGN", (0, 0), (-1, -1), "MIDDLE"),
        (
            "LINEBELOW", (0, 0), (-1, 0),
            0.5, colors.HexColor("#CBD5E1")
        )
    ]))

    return table

# ------------------------------------------------------------
# 12. CREATE PDF DOCUMENT
# ------------------------------------------------------------

doc = SimpleDocTemplate(
    str(pdf_path),
    pagesize=A4,
    rightMargin=40,
    leftMargin=40,
    topMargin=42,
    bottomMargin=42,
    title="Supermarket Sales Analysis Report",
    author="Data Analysis Portfolio"
)

story = []

# Title
story.append(para(
    "SUPERMARKET SALES ANALYSIS",
    "ReportTitle"
))

story.append(para(
    "Data Analysis Portfolio | Project 1"
))

story.append(para(
    f"Analysis period: {start_date} to {end_date}",
    "ReportNote"
))

story.append(Spacer(1, 16))

# Executive summary
story.append(section("1. Executive Summary"))

executive_summary = (
    f"This project analyzes {total_records:,} simulated "
    f"supermarket product-line sales records. Total "
    f"recorded sales after discounts were "
    f"INR {total_revenue:,.2f}. "
    f"{best_category['Category']} contributed the "
    f"largest category revenue share "
    f"({best_category['SharePercent']:.1f}%). "
    f"{best_product['Product']} was the highest-revenue "
    f"product. The strongest month was "
    f"{best_month['Month']}."
)

story.append(para(executive_summary))

# KPI summary
story.append(section("2. Key Performance Indicators"))

kpi_rows = [
    ["Metric", "Value"],
    ["Total Records", f"{total_records:,}"],
    ["Unique Customer IDs", f"{unique_customers:,}"],
    ["Units Sold", f"{total_units:,}"],
    ["Recorded Sales", f"INR {total_revenue:,.2f}"],
    ["Gross Sales", f"INR {gross_sales:,.2f}"],
    ["Discount Amount", f"INR {total_discount:,.2f}"],
    ["Average Product-Line Sales", f"INR {average_sales:,.2f}"],
    ["Median Product-Line Sales", f"INR {median_sales:,.2f}"]
]

story.append(make_table(kpi_rows, [235, 270]))

# Methodology
story.append(section("3. Analysis Methodology"))

story.append(para(
    "The dataset was loaded using Pandas. Exploratory "
    "analysis examined column types, missing values, "
    "duplicates, distributions and summary statistics. "
    "Category, product and monthly revenue were calculated "
    "using groupby aggregation. Visualizations were "
    "produced using Matplotlib and Seaborn."
))

# Category table
story.append(section("4. Category Performance"))

category_rows = [[
    "Category", "Revenue (INR)", "Units", "Share"
]]

for _, row in category_summary.iterrows():
    category_rows.append([
        str(row["Category"]),
        f"{row['Revenue']:,.0f}",
        f"{row['Units']:,.0f}",
        f"{row['SharePercent']:.1f}%"
    ])

story.append(make_table(
    category_rows, [150, 145, 90, 120]
))

# Monthly table
story.append(section("5. Monthly Sales Performance"))

month_rows = [[
    "Month", "Revenue (INR)", "Records", "Growth"
]]

for _, row in monthly_summary.iterrows():
    growth = (
        f"{row['GrowthPercent']:+.1f}%"
        if pd.notna(row["GrowthPercent"])
        else "N/A"
    )

    month_rows.append([
        row["Month"],
        f"{row['Revenue']:,.0f}",
        f"{row['Records']:,.0f}",
        growth
    ])

story.append(make_table(
    month_rows, [130, 155, 95, 125]
))

# Findings
story.append(section("6. Key Business Findings"))

for i, insight in enumerate(insights, 1):
    story.append(para(f"{i}. {insight}"))

# Recommendations
story.append(section("7. Recommendations"))

for i, recommendation in enumerate(recommendations, 1):
    story.append(para(f"{i}. {recommendation}"))

# Limitations
story.append(section("8. Limitations"))

limitations = [
    "The dataset is simulated and is intended for learning.",
    "Each transaction ID represents one recorded product line.",
    "No cost data is available to calculate profit margins.",
    "Revenue differences do not prove promotional effects.",
    "No transaction-hour data is available for peak-hour analysis.",
    "The analysis does not establish causal relationships."
]

for item in limitations:
    story.append(para("- " + item))

# ------------------------------------------------------------
# 13. EMBED STEP 9 CHARTS IN THE PDF
# ------------------------------------------------------------

story.append(PageBreak())
story.append(section("9. Data Visualization Appendix"))

chart_files = [
    ("Category Revenue", "01_category_revenue.png"),
    ("Top Products", "02_top_products.png"),
    ("Monthly Revenue", "03_monthly_revenue.png"),
    ("Daily Sales Trend", "04_daily_revenue_trend.png"),
    ("Payment Method Heatmap", "05_payment_heatmap.png"),
    ("Sales Distribution", "06_category_boxplot.png")
]

charts_added = 0

for title, filename in chart_files:
    image_path = chart_dir / filename

    if not image_path.exists():
        print("Chart not found:", filename)
        continue

    chart_image = Image(str(image_path))

    width = chart_image.imageWidth
    height = chart_image.imageHeight

    scale = min(480 / width, 235 / height)

    chart_image.drawWidth = width * scale
    chart_image.drawHeight = height * scale

    story.append(KeepTogether([
        section(title),
        chart_image,
        Spacer(1, 12)
    ]))

    charts_added += 1

# ------------------------------------------------------------
# 14. ADD PAGE NUMBERS AND BUILD REPORT
# ------------------------------------------------------------

def add_page_number(canvas, document):
    canvas.saveState()

    canvas.setFont("Helvetica", 8)
    canvas.setFillColor(colors.grey)

    canvas.drawString(
        40, 25,
        "Supermarket Sales Analysis | Simulated Dataset"
    )

    canvas.drawRightString(
        A4[0] - 40,
        25,
        f"Page {document.page}"
    )

    canvas.restoreState()

doc.build(
    story,
    onFirstPage=add_page_number,
    onLaterPages=add_page_number
)

# ------------------------------------------------------------
# 15. VERIFY GENERATED FILES
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("STEP 10 COMPLETED")
print("=" * 60)

print("PDF Report:", pdf_path)
print("Business Insights:", insights_path)

print("Charts embedded:", charts_added, "of 6")

assert pdf_path.exists()
assert pdf_path.stat().st_size > 0

print("\nPDF REPORT GENERATED SUCCESSFULLY!")
```
